```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display, HTML
import os
import re
from pathlib import Path
from itertools import combinations

# Configuración para mostrar todas las columnas
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
pd.set_option('display.width', None)

career_code_mapping = {
    'CIV-PLS13': 'Ing. Civil', 'CIV-PLS23': 'Ing. Civil', 'INT9CONSTR': 'Ing. Civil',
    'INT9TRANSP': 'Ing. Civil', 'INT9ORTERR': 'Ing. Civil', 'INT9SANEHI': 'Ing. Civil',
    'ELE-PLS13': 'Ing. Electrónica', 'ELE-PLS23': 'Ing. Electrónica',
    'INT9ELECTR': 'Ing. Electrónica', 'INT9SDIGYT': 'Ing. Electrónica',
    'MCT-PLS13': 'Ing. Mecatrónica', 'MCT-PLS23': 'Ing. Mecatrónica', 'MCT9-OPT': 'Ing. Mecatrónica',
    'IND-PLS13': 'Ing. Industrial', 'IND-PLS23': 'Ing. Industrial',
    'INT9G-ECO': 'Ing. Industrial', 'INT9-PROYT': 'Ing. Industrial',
    'CGF-PLS13': 'Ing. Geográfica', 'CGF-PLS23': 'Ing. Geográfica', 'INT9RNYMA': 'Ing. Geográfica',
    'MEC-PLS13': 'Ing. Mecánica', 'MEC-PLS23': 'Ing. Mecánica',
    'INT9MECANI': 'Ing. Mecánica', 'MEC9-OPT': 'Ing. Mecánica',
    'ECA-PLS13': 'Ing. Electromecánica', 'ECA-PLS23': 'Ing. Electromecánica', 'ECA9-OPT': 'Ing. Electromecánica'
}

# =============================================================================
# 1. CARGA Y CONCATENACIÓN DE DATOS
# =============================================================================

def cargar_y_concatenar(PATH, files):
    """
    Carga y concatena todos los archivos xls
    """
    datos_concatenados = pd.DataFrame()
    
    for i, filename in enumerate(files, 1):
        print(f"📄 Procesando archivo {i}/{len(files)}: {filename}")
        
        try:
            ruta_archivo = Path(PATH) / filename
            lectura = pd.read_excel(ruta_archivo, sheet_name="Sheet1", skiprows=0)
            
            match = re.search(r"_(\d{4})_ciclo_(\d)_", filename)
            if match:
                lectura['Anho'] = int(match.group(1))
                lectura['Semestre'] = int(match.group(2))
            else:
                print(f"  ⚠️ No se pudo extraer año/semestre de {filename}")
                continue
            
            lectura['Archivo_Origen'] = filename
            
            datos_concatenados = pd.concat([datos_concatenados, lectura], axis=0, ignore_index=True)
            print(f"  ✅ Registros agregados: {len(lectura):,} | Total acumulado: {len(datos_concatenados):,}")
            
        except Exception as e:
            print(f"  ❌ Error al procesar {filename}: {e}")
    
    print(f"\n✅ Total de registros concatenados: {len(datos_concatenados):,}")
    return datos_concatenados

# =============================================================================
# 2. ANÁLISIS DESCRIPTIVO GENERAL
# =============================================================================

def analisis_descriptivo_general(df):
    """
    Realiza un análisis descriptivo completo de los datos
    """
    print("="*80)
    print("📊 ANÁLISIS DESCRIPTIVO GENERAL DE LOS DATOS")
    print("="*80)
    
    # 2.1. Información básica del dataset
    print("\n" + "="*80)
    print("1. INFORMACIÓN BÁSICA DEL DATASET")
    print("="*80)
    
    print(f"📌 Total de registros: {len(df):,}")
    print(f"📌 Total de columnas: {len(df.columns)}")
    print(f"📌 Total de estudiantes únicos: {df['ALUMNO_ID'].nunique():,}")
    print(f"📌 Total de asignaturas únicas: {df['Asignatura'].nunique():,}")
    print(f"📌 Rango de años: {df['Anho'].min()} - {df['Anho'].max()}")
    print(f"📌 Semestres disponibles: {sorted(df['Semestre'].unique())}")
    
    # 2.2. Estructura de los datos
    print("\n" + "="*80)
    print("2. ESTRUCTURA DE LOS DATOS")
    print("="*80)
    
    print("\n📋 Primeras 5 filas:")
    display(df.head())
    
    print("\n📋 Últimas 5 filas:")
    display(df.tail())
    
    print("\n📋 Tipos de datos por columna:")
    print(df.dtypes)
    
    # 2.3. Estadísticas descriptivas
    print("\n" + "="*80)
    print("3. ESTADÍSTICAS DESCRIPTIVAS")
    print("="*80)
    
    # Columnas numéricas
    columnas_numericas = df.select_dtypes(include=[np.number]).columns
    print("\n📊 Estadísticas de columnas numéricas:")
    display(df[columnas_numericas].describe())
    
    # Columnas categóricas
    columnas_categoricas = df.select_dtypes(include=['object']).columns
    print("\n📊 Columnas categóricas:")
    for col in columnas_categoricas[:10]:  # Mostrar primeras 10
        print(f"\n{col}:")
        print(f"  - Valores únicos: {df[col].nunique()}")
        print(f"  - Más frecuente: {df[col].mode().iloc[0] if not df[col].mode().empty else 'N/A'}")
        print(f"  - Frecuencia: {df[col].value_counts().iloc[0] if not df[col].empty else 0}")
    
    # 2.4. Análisis de datos faltantes
    print("\n" + "="*80)
    print("4. ANÁLISIS DE DATOS FALTANTES")
    print("="*80)
    
    missing_data = df.isnull().sum()
    missing_percent = (missing_data / len(df)) * 100
    missing_df = pd.DataFrame({
        'Missing_Count': missing_data,
        'Missing_Percent': missing_percent
    })
    missing_df = missing_df[missing_df['Missing_Count'] > 0].sort_values('Missing_Count', ascending=False)
    
    if len(missing_df) > 0:
        print(f"\n🔍 Columnas con datos faltantes ({len(missing_df)} columnas):")
        display(missing_df)
    else:
        print("✅ No hay datos faltantes en el dataset")
    
    # 2.5. Análisis de duplicados
    print("\n" + "="*80)
    print("5. ANÁLISIS DE REGISTROS DUPLICADOS")
    print("="*80)
    
    duplicados = df.duplicated().sum()
    print(f"📌 Registros duplicados: {duplicados:,} ({duplicados/len(df)*100:.2f}%)")
    
    if duplicados > 0:
        print("\n🔍 Ejemplo de registros duplicados:")
        display(df[df.duplicated(keep=False)].head(5))
    
    # 2.6. Análisis por Año y Semestre
    print("\n" + "="*80)
    print("6. DISTRIBUCIÓN POR AÑO Y SEMESTRE")
    print("="*80)
    
    year_semester_counts = df.groupby(['Anho', 'Semestre']).size().reset_index(name='Registros')
    print("\n📊 Distribución de registros por año y semestre:")
    display(year_semester_counts)
    
    # Crear tabla pivote para mejor visualización
    pivot_counts = df.pivot_table(index='Anho', columns='Semestre', aggfunc='size', fill_value=0)
    print("\n📊 Tabla pivote de registros por año y semestre:")
    display(pivot_counts)
    
    # 2.7. Análisis de Asignaturas
    print("\n" + "="*80)
    print("7. ANÁLISIS DE ASIGNATURAS")
    print("="*80)
    
    # Top 10 asignaturas con más registros
    top_asignaturas = df['Asignatura'].value_counts().head(10)
    print("\n📚 Top 10 asignaturas con más registros:")
    display(top_asignaturas)
    
    # Asignaturas por nivel
    if 'Cod.Curso' in df.columns:
        asignaturas_por_nivel = df.groupby('Cod.Curso')['Asignatura'].nunique().sort_index()
        print("\n📚 Número de asignaturas por nivel:")
        print(asignaturas_por_nivel)
    
    # 2.8. Análisis de Carreras
    print("\n" + "="*80)
    print("8. ANÁLISIS DE CARRERAS")
    print("="*80)
    
    # Mapear nombre de carrera a partir de Cod.Car.Sec
    if 'Cod.Car.Sec' in df.columns:
        df['Carrera_Nombre'] = df['Cod.Car.Sec'].astype(str).str.strip().map(career_code_mapping)
        carreras_counts = df['Carrera_Nombre'].value_counts(dropna=False)
        print("\n🎓 Distribución de registros por carrera:")
        display(carreras_counts)

        if 'ALUMNO_ID' in df.columns:
            alumnos_por_carrera = (
                df.dropna(subset=['Carrera_Nombre'])
                .groupby('Carrera_Nombre')['ALUMNO_ID']
                .nunique()
                .sort_values(ascending=False)
            )
            print("\n👥 Distribución de alumnos únicos por carrera:")
            display(alumnos_por_carrera)

            carreras_por_alumno = (
                df.dropna(subset=['Carrera_Nombre'])
                .groupby('ALUMNO_ID')['Carrera_Nombre']
                .agg(lambda x: sorted(set(x)))
            )

            alumnos_multi_carrera = carreras_por_alumno[carreras_por_alumno.apply(len) > 1]
            print(f"\n🔁 Alumnos en más de una carrera: {len(alumnos_multi_carrera):,}")

            pares_carreras = [
                par
                for carreras in alumnos_multi_carrera
                for par in combinations(carreras, 2)
            ]

            if pares_carreras:
                ranking_carreras_compartidas = (
                    pd.Series(pares_carreras)
                    .value_counts()
                    .rename_axis('Par_Carreras')
                    .reset_index(name='Alumnos_Compartidos')
                )
                ranking_carreras_compartidas['Par_Carreras'] = ranking_carreras_compartidas['Par_Carreras'].apply(
                    lambda x: f"{x[0]} + {x[1]}"
                )

                print("\n🤝 Ranking de pares de carreras con más alumnos compartidos (Top 15):")
                display(ranking_carreras_compartidas.head(15))
            else:
                ranking_carreras_compartidas = pd.DataFrame(columns=['Par_Carreras', 'Alumnos_Compartidos'])
                print("\n🤝 No se encontraron alumnos con combinación de carreras.")
        
        # Verificar que son 7 carreras
        print(f"\n📌 Total de carreras únicas: {df['Carrera_Nombre'].nunique()}")
        print(f"📌 Carreras encontradas:")
        for carrera in sorted(df['Carrera_Nombre'].dropna().unique()):
            print(f"  - {carrera}")
    
    # 2.9. Análisis de Aprobación
    print("\n" + "="*80)
    print("9. ANÁLISIS DE APROBACIÓN")
    print("="*80)
    
    if 'Aprobado' in df.columns:
        aprobacion_counts = df['Aprobado'].value_counts()
        print("\n✅ Estado de aprobación:")
        display(aprobacion_counts)
        print(f"\n📊 Tasa de aprobación: {aprobacion_counts.get('S', 0)/len(df)*100:.2f}%")
        
        # Aprobación por año
        aprobacion_por_anho = df.groupby('Anho')['Aprobado'].value_counts(normalize=True).unstack()
        aprobacion_por_anho = aprobacion_por_anho * 100
        print("\n📈 Tasa de aprobación por año:")
        display(aprobacion_por_anho)
    
    # 2.10. Análisis de Notas
    print("\n" + "="*80)
    print("10. ANÁLISIS DE NOTAS")
    print("="*80)
    
    if 'Nota.Final' in df.columns:
        print(f"\n📊 Estadísticas de Nota Final:")
        print(df['Nota.Final'].describe())
        
        
    return {
        'df_info': df.info(),
        'missing_data': missing_df,
        'year_semester_counts': year_semester_counts,
        'carreras_counts': carreras_counts if 'Cod.Car.Sec' in df.columns else None,
        'alumnos_por_carrera': alumnos_por_carrera if 'Cod.Car.Sec' in df.columns and 'ALUMNO_ID' in df.columns else None,
        'alumnos_multi_carrera': len(alumnos_multi_carrera) if 'Cod.Car.Sec' in df.columns and 'ALUMNO_ID' in df.columns else 0,
        'ranking_carreras_compartidas': ranking_carreras_compartidas if 'Cod.Car.Sec' in df.columns and 'ALUMNO_ID' in df.columns else None,
        'aprobacion_counts': aprobacion_counts if 'Aprobado' in df.columns else None
    }

# =============================================================================
# 3. VISUALIZACIONES
# =============================================================================

def visualizar_datos(df):
    """
    Crea visualizaciones principales de los datos
    """
    print("\n" + "="*80)
    print("📈 VISUALIZACIONES DE DATOS")
    print("="*80)
    
    fig, axes = plt.subplots(2, 3, figsize=(18, 10))
    
    # 1. Distribución por año
    if 'Anho' in df.columns:
        df['Anho'].value_counts().sort_index().plot(kind='bar', ax=axes[0,0], color='skyblue')
        axes[0,0].set_title('Distribución de Registros por Año')
        axes[0,0].set_xlabel('Año')
        axes[0,0].set_ylabel('Número de Registros')
        axes[0,0].tick_params(axis='x', rotation=45)
    
    # 2. Distribución de registros por carrera
    if 'Carrera_Nombre' in df.columns:
        df['Carrera_Nombre'].value_counts().plot(kind='bar', ax=axes[0,1], color='lightgreen')
        axes[0,1].set_title('Registros por Carrera')
        axes[0,1].set_xlabel('Carrera')
        axes[0,1].set_ylabel('Número de Registros')
        axes[0,1].tick_params(axis='x', rotation=45)

    # 3. Comparación: registros vs alumnos únicos por carrera
    if 'Carrera_Nombre' in df.columns and 'ALUMNO_ID' in df.columns:
        comparacion_carreras = pd.concat([
            df['Carrera_Nombre'].value_counts(dropna=False).rename('Registros'),
            df.dropna(subset=['Carrera_Nombre']).groupby('Carrera_Nombre')['ALUMNO_ID'].nunique().rename('Alumnos_Unicos')
        ], axis=1).fillna(0)

        comparacion_carreras = comparacion_carreras.sort_values('Registros', ascending=False)
        comparacion_carreras.plot(kind='bar', ax=axes[1,1], color=['#66c2a5', '#fc8d62'])
        axes[1,1].set_title('Registros vs Alumnos Únicos por Carrera')
        axes[1,1].set_xlabel('Carrera')
        axes[1,1].set_ylabel('Cantidad')
        axes[1,1].tick_params(axis='x', rotation=45)
        axes[1,1].legend(['Registros', 'Alumnos Únicos'])
    
    # 4. Estado de aprobación
    if 'Aprobado' in df.columns:
        df['Aprobado'].value_counts().plot(kind='pie', ax=axes[0,2], autopct='%1.1f%%', colors=['lightcoral', 'lightblue'])
        axes[0,2].set_title('Estado de Aprobación')
        axes[0,2].set_ylabel('')
    
  
    
    # 5. Top 10 asignaturas
    df['Asignatura'].value_counts().head(10).plot(kind='barh', ax=axes[1,2], color='purple')
    axes[1,2].set_title('Top 10 Asignaturas con más Registros')
    axes[1,2].set_xlabel('Número de Registros')

    # Ocultar subplot no utilizado
    axes[1,0].axis('off')
    
    plt.tight_layout()
    plt.show()
    
    # Visualización adicional: Mapa de calor de datos faltantes
    print("\n📊 Mapa de calor de datos faltantes:")
    plt.figure(figsize=(12, 8))
    sns.heatmap(df.isnull(), cbar=True, yticklabels=False, cmap='viridis')
    plt.title('Mapa de Calor de Datos Faltantes')
    plt.show()

# =============================================================================
# 4. EJECUCIÓN PRINCIPAL
# =============================================================================

def main():
    """
    Función principal que ejecuta todo el análisis
    """
    base_dir = Path.cwd()
    posibles_dirs = [
        base_dir / "Datos_Nuevos",
        base_dir / "Datos Nuevos",
        base_dir / "TP_2" / "Datos_Nuevos",
        base_dir / "TP_2" / "Datos Nuevos",
    ]
    datos_dir = next((p for p in posibles_dirs if p.exists()), None)
    
    if datos_dir is None:
        print("❌ No se encontró la carpeta de datos")
        return
    
    files = sorted([f.name for f in datos_dir.iterdir() if f.is_file() and f.suffix.lower() in {'.xls', '.xlsx'}])
    
    if not files:
        print(f"❌ No se encontraron archivos XLS en {datos_dir}")
        return
    
    print(f"🔍 Encontrados {len(files)} archivos XLS para procesar")
    print(f"📂 Ruta: {datos_dir}")
    
    # 1. Cargar y concatenar datos
    print("\n" + "="*80)
    print("🔄 PROCESO DE CARGA Y CONCATENACIÓN")
    print("="*80)
    
    datos = cargar_y_concatenar(datos_dir, files)
    
    if len(datos) == 0:
        print("❌ No se pudieron cargar datos")
        return
    
    # 2. Análisis descriptivo general
    print("\n" + "="*80)
    print("📊 ANÁLISIS DESCRIPTIVO GENERAL")
    print("="*80)
    
    resultados = analisis_descriptivo_general(datos)
    
    # 3. Visualizaciones
    visualizar_datos(datos)
    
    # 4. Guardar datos procesados
    datos.to_csv("datos_concatenados.csv", index=False)
    print("\n✅ Datos concatenados guardados en 'datos_concatenados.csv'")
    
    # 5. Resumen final
    print("\n" + "="*80)
    print("📋 RESUMEN FINAL")
    print("="*80)
    
    print(f"""
    ✅ PROCESO COMPLETADO EXITOSAMENTE
    
    📊 Estadísticas del dataset:
    • Registros totales: {len(datos):,}
    • Estudiantes únicos: {datos['ALUMNO_ID'].nunique():,}
    • Asignaturas únicas: {datos['Asignatura'].nunique():,}
    • Años académicos: {datos['Anho'].min()} - {datos['Anho'].max()}
    • Semestres: {', '.join(map(str, sorted(datos['Semestre'].unique())))}
    • Carreras: {datos['Carrera_Nombre'].nunique() if 'Carrera_Nombre' in datos.columns else 'N/A'}
    
    📁 Archivos generados:
    • datos_concatenados.csv
    """)
    
    return datos, resultados

# =============================================================================
# EJECUTAR EL ANÁLISIS
# =============================================================================

if __name__ == "__main__":
    datos_procesados, resultados_analisis = main()
```

```python
# =============================================================================
# ANÁLISIS ESTADÍSTICO Y RESPUESTAS A LAS 6 PREGUNTAS
# =============================================================================

import pandas as pd
import numpy as np

# Selección del dataframe disponible
df_analisis = None
for name in ['df_clean', 'datos_procesados', 'datos', 'df_concat', 'df_raw', 'df']:
    if name in globals() and isinstance(globals()[name], pd.DataFrame):
        df_analisis = globals()[name]
        print(f"📌 Usando dataframe de memoria: '{name}'")
        break

if df_analisis is None:
    df_analisis = pd.read_csv('datos_concatenados.csv')
    print("📌 Cargado desde 'datos_concatenados.csv'")

print(f"📊 Total de registros analizados: {len(df_analisis):,}\n")

# Mapeo de carreras si no se encuentra definido
carrera_dict = {
    'C': 'Ing. Civil', 'E': 'Ing. Electrónica', 'I': 'Ing. Industrial',
    'M': 'Ing. Electromecánica', 'G': 'Ing. Geográfica',
    'K': 'Ing. Mecatrónica', 'L': 'Ing. Mecánica'
}
if 'Cod.Car.Sec' in df_analisis.columns and 'Carrera_Nombre' not in df_analisis.columns:
    df_analisis['Carrera_Nombre'] = df_analisis['Cod.Car.Sec'].astype(str).str.strip().map(carrera_dict)

# Conversión explícita de columnas numéricas
p1 = pd.to_numeric(df_analisis['Primer.Par'], errors='coerce')
p2 = pd.to_numeric(df_analisis['Segundo.Par'], errors='coerce')
nota_num = pd.to_numeric(df_analisis['Nota.Final'], errors='coerce')
firma_num = pd.to_numeric(df_analisis['Firma'], errors='coerce')
firmacalc_num = pd.to_numeric(df_analisis['FirmaCalculada'], errors='coerce') if 'FirmaCalculada' in df_analisis.columns else pd.Series(np.nan, index=df_analisis.index)
is_aprobado = df_analisis['Aprobado'].astype(str).str.strip().str.upper().isin(['S', 'SI', '1', '1.0'])
total_est = df_analisis['ALUMNO_ID'].nunique()

print("="*80)
print("1. ¿CUÁNTOS ESTUDIANTES PRESENTAN REGISTROS EN LOS TRES CICLOS DEL CSV?")
print("="*80)
df_analisis['Ciclo'] = df_analisis['Anho'].astype(str) + '-' + df_analisis['Semestre'].astype(str)
ciclos = sorted(df_analisis['Ciclo'].unique())
est_por_ciclo = df_analisis.groupby('ALUMNO_ID')['Ciclo'].nunique()
est_3_ciclos = (est_por_ciclo == len(ciclos)).sum()
print(f"• Ciclos analizados: {ciclos}")
print(f"• Estudiantes únicos en los 3 ciclos: {est_3_ciclos:,} de {total_est:,} ({est_3_ciclos/total_est*100:.2f}%)\n")

print("="*80)
print("2. ¿CUÁNTAS PERSONAS/REGISTROS TIENEN CALIFICACIÓN FINAL (Nota.Final)?")
print("="*80)
con_nota = df_analisis['Nota.Final'].notnull()
cant_reg_nota = con_nota.sum()
cant_pers_nota = df_analisis[con_nota]['ALUMNO_ID'].nunique()
print(f"• Registros con Nota.Final (!= NaN): {cant_reg_nota:,} de {len(df_analisis):,} ({cant_reg_nota/len(df_analisis)*100:.2f}%)")
print(f"• Personas únicas con Nota.Final: {cant_pers_nota:,} de {total_est:,} ({cant_pers_nota/total_est*100:.2f}%)\n")

print("="*80)
print("3. ¿CUÁNTOS TIENEN PROCESO (FirmaCalculada / Firma)?")
print("="*80)
proc_gt0 = (firma_num > 0) | (firmacalc_num > 0)
proc_ge50 = (firma_num >= 50) | (firmacalc_num >= 50)
print(f"• Registros con proceso registrado (> 0 pts): {proc_gt0.sum():,} ({proc_gt0.mean()*100:.2f}%)")
print(f"• Personas únicas con proceso (> 0 pts): {df_analisis[proc_gt0]['ALUMNO_ID'].nunique():,} ({df_analisis[proc_gt0]['ALUMNO_ID'].nunique()/total_est*100:.2f}%)")
print(f"• Registros con Firma alcanzada (>= 50 pts): {proc_ge50.sum():,} ({proc_ge50.mean()*100:.2f}%)")
print(f"• Personas únicas con Firma alcanzada (>= 50 pts): {df_analisis[proc_ge50]['ALUMNO_ID'].nunique():,} ({df_analisis[proc_ge50]['ALUMNO_ID'].nunique()/total_est*100:.2f}%)\n")

print("="*80)
print("4. ¿CUÁL ES LA TASA DE APROBACIÓN GLOBAL Y POR CARRERA?")
print("="*80)
tasa_global = is_aprobado.mean() * 100
print(f"• Tasa de Aprobación Global: {tasa_global:.2f}% ({is_aprobado.sum():,} / {len(df_analisis):,})")
print("• Tasa de Aprobación por Carrera:")
if 'Carrera_Nombre' in df_analisis.columns:
    for car, grp in df_analisis.groupby('Carrera_Nombre'):
        apr = grp['Aprobado'].astype(str).str.strip().str.upper().isin(['S', 'SI', '1', '1.0'])
        print(f"   - {car:<25}: {apr.mean()*100:6.2f}% ({apr.sum():,} / {len(grp):,})")
print()

print("="*80)
print("5. ¿CÓMO SE COMPARAN LAS NOTAS MEDIAS DEL 1ER Y 2DO PARCIAL SEGÚN CONDICIÓN FINAL?")
print("="*80)
df_p = pd.DataFrame({'Condicion': np.where(is_aprobado, 'Aprobado', 'No Aprobado'), 'P1': p1, 'P2': p2})
m_apr_p1 = df_p[df_p['Condicion']=='Aprobado']['P1'].mean()
m_apr_p2 = df_p[df_p['Condicion']=='Aprobado']['P2'].mean()
m_no_p1 = df_p[df_p['Condicion']=='No Aprobado']['P1'].mean()
m_no_p2 = df_p[df_p['Condicion']=='No Aprobado']['P2'].mean()
print(f"• Estudiantes APROBADOS:")
print(f"   - Media 1er Parcial: {m_apr_p1:.2f} pts (N = {df_p[df_p['Condicion']=='Aprobado']['P1'].count():,})")
print(f"   - Media 2do Parcial: {m_apr_p2:.2f} pts (N = {df_p[df_p['Condicion']=='Aprobado']['P2'].count():,})")
print(f"• Estudiantes NO APROBADOS:")
print(f"   - Media 1er Parcial: {m_no_p1:.2f} pts (N = {df_p[df_p['Condicion']=='No Aprobado']['P1'].count():,})")
print(f"   - Media 2do Parcial: {m_no_p2:.2f} pts (N = {df_p[df_p['Condicion']=='No Aprobado']['P2'].count():,})\n")

print("="*80)
print("6. ¿CUÁL ES LA TASA DE ABANDONO / INASISTENCIA TOTAL A PARCIALES?")
print("="*80)
sin_p1 = p1.isna() | (p1 == 0)
sin_p2 = p2.isna() | (p2 == 0)
abandono_total = sin_p1 & sin_p2
sin_al_menos_uno = sin_p1 | sin_p2
print(f"• Inasistencia al 1er Parcial (NaN / 0): {sin_p1.mean()*100:.2f}% ({sin_p1.sum():,} de {len(df_analisis):,})")
print(f"• Inasistencia al 2do Parcial (NaN / 0): {sin_p2.mean()*100:.2f}% ({sin_p2.sum():,} de {len(df_analisis):,})")
print(f"• ABANDONO TOTAL (Inasistencia a AMBOS parciales): {abandono_total.mean()*100:.2f}% ({abandono_total.sum():,} de {len(df_analisis):,})")
print(f"• Inasistencia a al menos un parcial: {sin_al_menos_uno.mean()*100:.2f}% ({sin_al_menos_uno.sum():,} de {len(df_analisis):,})\n")

# Resumen final tabular para uso académico y presentación
resumen_final = pd.DataFrame({
    'Pregunta': [
        '1. Estudiantes en 3 ciclos',
        '2. Registros con Nota.Final',
        '2. Personas con Nota.Final',
        '3. Registros con proceso',
        '3. Personas con proceso',
        '4. Tasa de aprobación global',
        '5. Media P1 aprobados',
        '5. Media P2 aprobados',
        '5. Media P1 no aprobados',
        '5. Media P2 no aprobados',
        '6. Abandono total',
        '6. Inasistencia a al menos un parcial'
    ],
    'Valor': [
        f"{est_3_ciclos:,} / {total_est:,} ({est_3_ciclos/total_est*100:.2f}%)",
        f"{cant_reg_nota:,} / {len(df_analisis):,} ({cant_reg_nota/len(df_analisis)*100:.2f}%)",
        f"{cant_pers_nota:,} / {total_est:,} ({cant_pers_nota/total_est*100:.2f}%)",
        f"{proc_gt0.sum():,} / {len(df_analisis):,} ({proc_gt0.mean()*100:.2f}%)",
        f"{df_analisis[proc_gt0]['ALUMNO_ID'].nunique():,} / {total_est:,} ({df_analisis[proc_gt0]['ALUMNO_ID'].nunique()/total_est*100:.2f}%)",
        f"{is_aprobado.mean()*100:.2f}%",
        f"{m_apr_p1:.2f} pts",
        f"{m_apr_p2:.2f} pts",
        f"{m_no_p1:.2f} pts",
        f"{m_no_p2:.2f} pts",
        f"{abandono_total.mean()*100:.2f}% ({abandono_total.sum():,} / {len(df_analisis):,})",
        f"{sin_al_menos_uno.mean()*100:.2f}% ({sin_al_menos_uno.sum():,} / {len(df_analisis):,})"
    ]
})

print("="*80)
print("RESUMEN FINAL DE RESULTADOS")
print("="*80)
print(resumen_final.to_string(index=False))

print("\nInterpretación breve:")
print(f"- La cohorte tiene {total_est:,} estudiantes únicos y {len(df_analisis):,} registros, con {est_3_ciclos:,} alumnos presentes en los 3 ciclos.")
print(f"- La aprobación global alcanza {is_aprobado.mean()*100:.2f}% y la diferencia de medias entre aprobados y no aprobados es clara en ambos parciales.")
print(f"- La inasistencia acumulada es significativa: {abandono_total.mean()*100:.2f}% no asistió a ambos parciales y {sin_al_menos_uno.mean()*100:.2f}% faltó a al menos uno.")
```
